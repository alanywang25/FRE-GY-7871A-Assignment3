## Methodological foundation

This notebook combines two papers. **Rigobon (2003), Identification through Heteroskedasticity**, shows how shifts in structural-shock variances can identify a simultaneous system when conventional exclusion restrictions or observable instruments are unavailable. **Rigobon and Sack (2003), The Effects of War Risk on U.S. Financial Markets**, applies that insight to a latent war-risk factor: the factor need not be observed or signed on each day, provided researchers can identify dates when its variance is unusually high.

Let daily changes in financial variables be `Delta x_t = D z_t + e_t`, where the first element of the unobserved factor vector `z_t` is Iran war risk. For a high-news regime H and matched low-news regime L, the key identifying restriction is that the relevant change in the reduced-form covariance matrix, `Delta Sigma = Sigma_H - Sigma_L`, is driven by a change in the variance of that war-risk factor. With the two-year Treasury-yield change as the normalized variable, covariance ratios and combined-instrument IV recover each market variable's loading on the latent factor.

The central assumptions are demanding: market loadings must remain stable across regimes; the Iran-war-risk shock must be orthogonal to other shocks; and variance changes in other shocks must not systematically differ between high- and low-news dates. The notebook reports two covariance-ratio estimators, combined IV, bootstrap intervals, and a covariance-shift diagnostic because agreement across these pieces of evidence is more informative than any single coefficient.

### How GKG substitutes for newspaper reading

The original war-risk paper identifies high-variance dates by reading newspapers and market commentary. Here, cached GDELT Global Knowledge Graph (GKG) files provide a reproducible structured-news analogue. Each daily file is screened for Iran-related and conflict-related themes, locations, entities, and URLs. The high-news regime is the 17 trading dates with the greatest GKG Iran-conflict intensity; nearby, lower-intensity trading days form the comparison regime. Table 1 records this event calendar and representative source URLs so the mechanical selection can be audited.

GKG is not a full-text archive: it provides structured metadata and tone rather than article headlines or semantic article-body analysis. Therefore, the GKG signal is a transparent proxy for the intensity of Iran-war coverage, not a direct observed measure of the probability, direction, or severity of war risk. The analysis deliberately uses it to define variance regimes rather than to claim that a one-unit score is a one-unit increase in risk.

### Estimands and interpretation

As in Rigobon and Sack, all Table 2 responses are normalized to a war-risk shock that lowers the two-year Treasury yield by 25 basis points. Equity, commodity, dollar, credit-relative-performance, and VIX responses are daily log-return or log-change effects under that normalization. Table 3 translates the covariance shift into an estimated incremental variance contribution. These are conditional, same-day sensitivities - not forecasts, investment recommendations, or definitive causal effects.

The final comparison section implements three alternatives: a high-minus-low event study, a continuous GKG-intensity regression, and a PCA GKG news factor. They make stronger assumptions that GKG coverage is a cardinal quantitative risk proxy, so they complement rather than replace identification through heteroskedasticity. Comparing signs, magnitudes, uncertainty, and cross-method agreement is a robustness exercise.

## Automated GKG download and structured NLP

Full daily GKG files are sizeable. The cache makes this resumable: already-downloaded files are never requested again. Set `DOWNLOAD=False` after a successful download. The parser only loads the URL, Themes, and V2Tone columns required for the event measure.

```python
DOWNLOAD = True
def gkg_path(day): return GKG_DIR / f'{day:%Y%m%d}.gkg.csv.zip'
def download_gkg(day, retries=3):
    path = gkg_path(day)
    if path.exists(): return path
    url = f'https://data.gdeltproject.org/gkg/{path.name}'
    for attempt in range(retries):
        try:
            req = Request(url, headers={'User-Agent':'Mozilla/5.0 academic replication'})
            with urlopen(req, timeout=180) as response: path.write_bytes(response.read())
            return path
        except HTTPError as err:
            if err.code == 404: return None
            if err.code != 429: raise
            time.sleep(60 * (attempt + 1))
        except URLError:
            time.sleep(60 * (attempt + 1))
    return None

days = pd.date_range(WAR_START, END, freq='D')
if DOWNLOAD:
    unavailable=[]
    for day in days:
        if download_gkg(day) is None: unavailable.append(day.strftime('%F'))
        time.sleep(2)
    print('Unavailable dates:', unavailable)

IRAN = 'IRAN'; CONFLICT = ('WAR','CONFLICT','MILITARY','TERROR','VIOLENCE','STRIKE','MISSILE','BOMB','CEASEFIRE','HORMUZ')
def tone_value(value):
    try: return float(str(value).split(',')[0])
    except (ValueError, IndexError): return np.nan

def read_gkg(day):
    path = gkg_path(day)

    if not path.exists():
        return None

    raw = pd.read_csv(
        path,
        sep="\t",
        header=None,
        compression="zip",
        dtype=str,
        on_bad_lines="skip",
        low_memory=False
    ).fillna("")

    n_cols = raw.shape[1]

    # Search all available GKG analytical fields, rather than one
    # version-specific “Themes” column.
    feature_end = min(n_cols, 16)
    features = raw.iloc[:, 4:feature_end].agg(" ".join, axis=1).str.upper()

    iran = features.str.contains("IRAN", regex=False)

    conflict_pattern = (
        "WAR|CONFLICT|MILITARY|TERROR|VIOLENCE|STRIKE|MISSILE|"
        "BOMB|CEASEFIRE|HORMUZ|RETALIATION|NUCLEAR"
    )
    conflict = features.str.contains(conflict_pattern, regex=True)

    # GKG 2.1 generally uses V2Tone at column 15.
    # Older GKG layouts generally use Tone at column 10.
    if n_cols > 15:
        tone = raw.iloc[:, 15].map(tone_value)
    elif n_cols > 10:
        tone = raw.iloc[:, 10].map(tone_value)
    else:
        tone = pd.Series(np.nan, index=raw.index)

    selected = iran & conflict
    representative_urls = raw.loc[selected, 4].replace('', np.nan).dropna()

    return {
        "date": day,
        "iran_docs": int(iran.sum()),
        "iran_conflict_docs": int(selected.sum()),
        "mean_tone": tone[selected].mean(),
        "negative_tone_docs": int((tone[selected] < 0).sum()),
        "representative_url": representative_urls.iloc[0] if len(representative_urls) else np.nan
    }

records = [r for d in days if (r := read_gkg(d)) is not None]
daily = pd.DataFrame(records).set_index('date').sort_index()
daily["war_news_intensity"] = (
    np.log1p(daily["iran_conflict_docs"])
    + 0.25 * np.log1p(daily["iran_docs"])
    + 0.10 * np.log1p(daily["negative_tone_docs"])
)
daily.to_csv(OUTPUT_DIR / 'iran_war_risk_gkg_news_nlp.csv')
daily.describe()
```

```python
# News-only high-variance regime and matched low-news controls.
trading = pd.DatetimeIndex(panel.dropna(subset=['DGS2_bp']).index).intersection(daily.index)
eligible = daily[(daily.iran_conflict_docs >= 1) & daily.index.isin(trading)].sort_values('war_news_intensity', ascending=False)
high = list(eligible.head(N_EVENTS).index); used=set(high); low=[]
for h in sorted(high):
    nearby=sorted((d for d in trading if d not in used), key=lambda d:(abs((d-h).days),d))
    choice=next(d for d in nearby if d not in daily.index or daily.loc[d,'war_news_intensity'] <= daily.war_news_intensity.median())
    low.append(choice); used.add(choice)
audit=pd.concat([daily.reindex(high).assign(regime='high'),daily.reindex(low).assign(regime='low')]).reset_index(names='date').sort_values(['regime','date'])
audit.to_csv(OUTPUT_DIR / 'iran_war_risk_gkg_event_audit.csv',index=False); audit
```

## Table 1. Dates of high variance of Iran war risk

This is the 2026 Iran-war counterpart to Table 1 in Rigobon and Sack (2003). Rather than manually assigning event descriptions from newspaper reading, high-variance dates are selected mechanically as the 17 highest GKG Iran-conflict-intensity trading dates. The table provides the date, the reproducible selection statistic, coverage counts, and a representative source document for manual event validation.

```python
# Replication-style event calendar. Run after the GKG parser and event-selection cell.
table1 = (audit.loc[audit.regime == 'high', ['date','war_news_intensity','iran_docs','iran_conflict_docs','mean_tone','representative_url']]
         .sort_values('date').reset_index(drop=True))
table1.insert(0, 'event_number', np.arange(1, len(table1)+1))
table1['selection_rule'] = 'Top-17 GKG Iran-conflict intensity trading date'
table1['event_description'] = 'GKG-selected high-variance Iran-war-news date; validate using representative URL'
table1 = table1[['event_number','date','event_description','war_news_intensity','iran_docs','iran_conflict_docs','mean_tone','representative_url','selection_rule']]
table1.to_csv(OUTPUT_DIR / 'iran_war_risk_table_1_high_variance_dates.csv', index=False)
display(table1)
print('Table 1 interpretation: these dates identify unusually intense Iran-conflict coverage, not necessarily unambiguously positive or negative war-risk shocks. Review representative URLs and amend event descriptions only with cited external evidence.')
```

```python
# Identification through heteroskedasticity; scale the latent factor to a 25 bp DGS2 decline.
def estimate(hdates, ldates):
    H,L=panel.reindex(hdates).dropna(),panel.reindex(ldates).dropna(); cols=H.columns.intersection(L.columns); H,L=H[cols],L[cols]; delta=H.cov()-L.cov(); rows=[]
    for y in cols.drop('DGS2_bp'):
        bx=delta.loc[y,'DGS2_bp']/delta.loc['DGS2_bp','DGS2_bp']; by=delta.loc[y,y]/delta.loc[y,'DGS2_bp']
        q=pd.concat([H[['DGS2_bp',y]].assign(s=1),L[['DGS2_bp',y]].assign(s=-1)]); X,Y=q[['DGS2_bp']].to_numpy(),q[y].to_numpy(); Z=np.column_stack([q.s*q.DGS2_bp,q.s*q[y]])
        P=Z@np.linalg.pinv(Z.T@Z)@Z.T; b=float(np.linalg.pinv(X.T@P@X)@(X.T@P@Y)); rows.append([y,bx,by,b,-25*b,abs(bx-by)])
    return pd.DataFrame(rows,columns=['variable','beta_x_IV','beta_y_IV','combined_IV','response_to_minus_25bp','ratio_gap']),delta

results,delta=estimate(high,low); draws=[]
for _ in range(N_BOOT):
    try: draws.append(estimate(rng.choice(high,len(high),True),rng.choice(low,len(low),True))[0].set_index('variable').response_to_minus_25bp)
    except np.linalg.LinAlgError: pass
draws=pd.DataFrame(draws); results=results.set_index('variable'); results['ci_2_5']=draws.quantile(.025); results['ci_97_5']=draws.quantile(.975); results['excludes_zero_95']=(results.ci_2_5>0)|(results.ci_97_5<0)
results.to_csv(OUTPUT_DIR / 'iran_war_risk_gkg_results.csv'); print('Covariance-shift singular values:',np.round(np.linalg.svd(delta.fillna(0),compute_uv=False),4)); results.sort_values('response_to_minus_25bp')
```

```python
# Automated, results-driven written analysis. Run after the event and estimation cells.
n_high = (audit.regime == 'high').sum(); n_low = (audit.regime == 'low').sum()
missing_low = audit.loc[audit.regime == 'low', 'iran_conflict_docs'].isna().sum()
significant = results[results.excludes_zero_95]
def direction(row):
    return 'increase' if row.response_to_minus_25bp > 0 else 'decrease'

print('## Results interpretation')
print(f'The GKG rule selected {n_high} high-news dates and {n_low} matched low-news dates.')
if missing_low:
    print(f'WARNING: {missing_low} low-news controls lack GKG coverage. Do not interpret estimates until the event-selection cell is rerun with the updated trading-day restriction.')
else:
    print('All selected high and low dates have GKG coverage; the event comparison is internally consistent.')

if significant.empty:
    print('No response has a 95% bootstrap interval excluding zero, so the evidence is statistically inconclusive in this short sample.')
else:
    print('Responses with 95% bootstrap intervals excluding zero:')
    for name, row in significant.iterrows():
        print(f'- {name}: {direction(row)} of {row.response_to_minus_25bp:.2f} for a 25 bp two-year-yield decline (95% CI {row.ci_2_5:.2f} to {row.ci_97_5:.2f}).')

large_gap = results[results.ratio_gap > results.ratio_gap.median()]
print(f'Estimator-agreement check: {len(large_gap)} of {len(results)} variables have an above-median gap between the two covariance-ratio estimates. Large gaps indicate sensitivity to identifying assumptions and should temper causal language.')
print('Economic reading: interpret only signs supported by intervals that exclude zero. In particular, do not infer a conventional risk-off pattern merely from an imprecise point estimate.')
```

```python
# Additional interpretation of the event and response tables. Run after estimation.
expected = {'SPY_pct':-1,'EFA_pct':-1,'EEM_pct':-1,'Brent_pct':1,'Gold_pct':1,'Dollar_pct':-1,'Credit_HYG_minus_IEF_pct':-1,'VIX_pct':1}
labels = {'SPY_pct':'US equities','EFA_pct':'developed ex-US equities','EEM_pct':'emerging-market equities','Brent_pct':'Brent crude','Gold_pct':'gold','Dollar_pct':'US dollar','Credit_HYG_minus_IEF_pct':'high-yield credit relative to Treasuries','VIX_pct':'implied equity volatility'}
table = results[['response_to_minus_25bp','ci_2_5','ci_97_5','ratio_gap','excludes_zero_95']].copy()
table['market']=[labels.get(v,v) for v in table.index]
table['expected_war_risk_direction']=['up' if expected[v]>0 else 'down' for v in table.index]
table['estimated_direction']=np.where(table.response_to_minus_25bp>0,'up','down')
table['interpretation']=np.where(~table.excludes_zero_95,'Imprecise: avoid directional inference',np.where([np.sign(table.loc[v,'response_to_minus_25bp'])==expected[v] for v in table.index],'Precise and consistent with risk-off benchmark','Precise but differs from risk-off benchmark'))
display(table[['market','expected_war_risk_direction','estimated_direction','response_to_minus_25bp','ci_2_5','ci_97_5','ratio_gap','interpretation']])
hi=audit.loc[audit.regime=='high','war_news_intensity'].median(); lo=audit.loc[audit.regime=='low','war_news_intensity'].median()
print(f'Event-table separation: median intensity is {hi:.2f} on high-news dates and {lo:.2f} on low-news dates.')
print('A large separation supports the volatility-regime design, but does not prove that other shocks had stable variances.')
print('Use combined_IV as the headline estimate, bootstrap intervals for precision, and ratio_gap to identify fragile estimates where the two covariance-ratio estimators disagree.')
print('A qualitative replication requires equity-down, Brent-up, dollar-down, high-yield-underperformance, and VIX-up signs with intervals excluding zero; otherwise characterize the result as divergent or inconclusive rather than causal.')
```

## Table 2. Estimated effects of Iran war risk on financial markets

This table is the Iran-2026 counterpart to Table 2 in Rigobon and Sack (2003). It reports the effect of a latent war-risk shock normalized to lower the two-year Treasury yield by 25 bp. The first two columns are the two covariance-ratio/IV estimators; the third is the combined-instrument IV estimate. Bootstrap intervals replace the paper's reported asymptotic significance statistics.

```python
# Replication-style Table 2. Run after the identification-through-heteroskedasticity estimation cell.
table2_labels = {'SPY_pct':'US equities (SPY)','EFA_pct':'Developed ex-US equities (EFA)','EEM_pct':'Emerging-market equities (EEM)','Brent_pct':'Brent crude futures','Gold_pct':'Gold futures','Dollar_pct':'US dollar index','Credit_HYG_minus_IEF_pct':'High-yield credit relative to Treasuries','VIX_pct':'Equity implied volatility (VIX)'}
table2_units = {'SPY_pct':'% log return','EFA_pct':'% log return','EEM_pct':'% log return','Brent_pct':'% log return','Gold_pct':'% log return','Dollar_pct':'% log return','Credit_HYG_minus_IEF_pct':'% relative return','VIX_pct':'% log change'}
table2 = results.copy()
table2['variable'] = [table2_labels.get(v,v) for v in table2.index]
table2['unit'] = [table2_units.get(v,'') for v in table2.index]
table2['2Y_instrument_effect'] = -25 * table2.beta_x_IV
table2['variable_instrument_effect'] = -25 * table2.beta_y_IV
table2['combined_IV_effect'] = table2.response_to_minus_25bp
table2['bootstrap_95pct_CI'] = table2.apply(lambda r: f'[{r.ci_2_5:.2f}, {r.ci_97_5:.2f}]', axis=1)
table2['significance'] = np.where(table2.excludes_zero_95, '95% CI excludes zero', 'Not precise at 95%')
table2 = table2[['variable','unit','2Y_instrument_effect','variable_instrument_effect','combined_IV_effect','bootstrap_95pct_CI','significance','ratio_gap']]
table2 = table2.sort_values('variable').reset_index(drop=True)
table2.to_csv(OUTPUT_DIR / 'iran_war_risk_table_2_financial_market_effects.csv', index=False)
display(table2)
print('Table 2 note: effects are conditional on the high-versus-low GKG regime identifying a war-risk-specific variance shift. Large ratio_gap values signal disagreement between the first two estimators; prioritize the combined-IV estimate only when this diagnostic is acceptable.')
```

## Table 3. Incremental variance attributable to Iran war risk

This table parallels the variance-accounting exercise in Rigobon and Sack (2003). With the two-year yield loading normalized to one, the high-minus-low covariance shift estimates an incremental war-risk variance, `ΔVar(z) = ΔVar(DGS2)`. For each variable, the implied incremental war-risk variance is `β² ΔVar(z)` and its reported share divides that quantity by the variable's total daily variance in the 2026 war-period sample. The decomposition is meaningful only if the covariance shift is primarily war-risk-specific and positive.

```python
# Replication-style Table 3. Run after the identification-through-heteroskedasticity estimation cell.
war_period = panel.loc[WAR_START:END].dropna(subset=['DGS2_bp'])
delta_var_2y = delta.loc['DGS2_bp','DGS2_bp']
rows = []
for variable in panel.columns:
    beta = 1.0 if variable == 'DGS2_bp' else results.loc[variable,'combined_IV']
    total_variance = war_period[variable].var(ddof=1)
    implied_war_variance = beta**2 * delta_var_2y
    share = 100 * implied_war_variance / total_variance if total_variance > 0 else np.nan
    rows.append([variable, beta, total_variance, implied_war_variance, share])
table3 = pd.DataFrame(rows, columns=['variable','war_risk_loading','total_daily_variance','incremental_war_risk_variance','incremental_variance_share_pct'])
table3['market'] = table3.variable.map({'DGS2_bp':'Two-year Treasury yield','SPY_pct':'US equities (SPY)','EFA_pct':'Developed ex-US equities (EFA)','EEM_pct':'Emerging-market equities (EEM)','Brent_pct':'Brent crude futures','Gold_pct':'Gold futures','Dollar_pct':'US dollar index','Credit_HYG_minus_IEF_pct':'High-yield credit relative to Treasuries','VIX_pct':'Equity implied volatility (VIX)'})
table3['diagnostic'] = np.where(delta_var_2y <= 0, 'Invalid: DGS2 high-low variance shift is non-positive', np.where(table3.incremental_variance_share_pct > 100, 'Warning: share exceeds 100%; assumptions/normalization strained', 'Interpretable conditional on identification assumptions'))
table3 = table3[['market','war_risk_loading','total_daily_variance','incremental_war_risk_variance','incremental_variance_share_pct','diagnostic']].sort_values('market').reset_index(drop=True)
table3.to_csv(OUTPUT_DIR / 'iran_war_risk_table_3_variance_decomposition.csv', index=False)
print(f'High-minus-low two-year-yield variance shift: {delta_var_2y:.4f} bp-squared.')
display(table3)
print('Table 3 interpretation: larger shares mean the estimated war-risk component accounts for more of a variable’s war-period volatility. A negative shift or shares above 100% are diagnostics against treating the decomposition as reliable.')
```

## Alternative war-risk measures and comparison

The heteroskedasticity estimator identifies exposure from a covariance shift between high- and low-news regimes, without treating the news score as a cardinal measure of risk. Three complementary approaches are implemented below: (1) a high-minus-low event study; (2) a continuous regression on standardized GKG news intensity; and (3) a PCA news factor built from GKG coverage features. Each alternative is calibrated to the same 25 bp decline in the two-year yield, making effect sizes comparable. Unlike identification through heteroskedasticity, however, these methods require the observed GKG signal to be a valid, linearly measured proxy for war risk and are more exposed to omitted confounding news.

```python
# Alternative estimators, all reported as the response associated with a 25 bp DGS2 decline.
news_market = panel.join(daily[['iran_docs','iran_conflict_docs','negative_tone_docs','war_news_intensity']], how='inner').dropna(subset=['DGS2_bp','war_news_intensity'])
def slope(y, x):
    valid = pd.concat([y, x], axis=1).dropna()
    if len(valid) < 10 or valid.iloc[:,1].var() == 0: return np.nan
    return np.linalg.lstsq(np.column_stack([np.ones(len(valid)), valid.iloc[:,1]]), valid.iloc[:,0], rcond=None)[0][1]
def calibrated_response(y, factor):
    b_y, b_x = slope(y, factor), slope(news_market['DGS2_bp'], factor)
    return np.nan if not np.isfinite(b_x) or abs(b_x) < 1e-8 else -25 * b_y / b_x

# (1) Mean high-minus-low event study, calibrated by the corresponding DGS2 mean difference.
H, L = panel.reindex(high).dropna(), panel.reindex(low).dropna()
event_scale = H.DGS2_bp.mean() - L.DGS2_bp.mean()
event_study = {v: np.nan if abs(event_scale) < 1e-8 else -25*(H[v].mean()-L[v].mean())/event_scale for v in panel.columns.drop('DGS2_bp')}

# (2) Continuous GKG intensity regression.
intensity_z = (news_market.war_news_intensity-news_market.war_news_intensity.mean())/news_market.war_news_intensity.std(ddof=0)
continuous = {v: calibrated_response(news_market[v], intensity_z) for v in panel.columns.drop('DGS2_bp')}

# (3) PCA factor from nonconstant standardized GKG features; orient it so greater conflict coverage is positive.
features = np.log1p(news_market[['iran_docs','iran_conflict_docs','negative_tone_docs']])
features = features.loc[:, features.std(ddof=0) > 0]
Z = (features-features.mean())/features.std(ddof=0)
pc1 = pd.Series(np.linalg.svd(Z.to_numpy(), full_matrices=False)[0][:,0], index=Z.index, name='gkg_pc1')
if pc1.corr(np.log1p(news_market.iran_conflict_docs)) < 0: pc1 = -pc1
pca_factor = {v: calibrated_response(news_market[v], pc1) for v in panel.columns.drop('DGS2_bp')}

comparison = pd.DataFrame({'identification_through_heteroskedasticity':results.combined_IV.mul(-25),'event_study_high_minus_low':pd.Series(event_study),'continuous_gkg_intensity':pd.Series(continuous),'pca_gkg_news_factor':pd.Series(pca_factor)})
comparison['sign_agreement_all_methods'] = comparison.apply(lambda row: len(set(np.sign(row.dropna()))) == 1, axis=1)
comparison.to_csv(OUTPUT_DIR / 'iran_war_risk_method_comparison.csv')
print(f'Event-study DGS2 mean difference: {event_scale:.2f} bp. Values close to zero make event-study calibration unstable.')
print(f'PCA features used: {list(features.columns)}. PC1 explained {np.linalg.svd(Z.to_numpy(), full_matrices=False)[1][0]**2 / (np.linalg.svd(Z.to_numpy(), full_matrices=False)[1]**2).sum():.1%} of standardized-feature variance.')
display(comparison)
```

```python
# Written comparison: this cell turns the displayed method-comparison table into guarded conclusions.
for variable, row in comparison.iterrows():
    available = row.iloc[:4].dropna()
    if len(available) < 2:
        print(f'{variable}: insufficient comparable estimates.')
    elif len(set(np.sign(available))) == 1:
        direction = 'increase' if available.iloc[0] > 0 else 'decrease'
        print(f'{variable}: all available methods imply a {direction}; this is directional robustness, not proof of causality.')
    else:
        print(f'{variable}: methods disagree in sign; report this sensitivity rather than selecting a preferred estimate.')
print('Method comparison: identification through heteroskedasticity is preferred when the regime-specific variance shift is credible, because it does not require a cardinal news-risk measure. The event study is most transparent but vulnerable to coincident news. Continuous intensity and PCA gain daily information but assume GKG coverage/thematic features are valid quantitative proxies and are most exposed to omitted variables.')
```
